```python
from HSA_GTMo_C_001_numpyro import calculate_partial_pressures
from HSA_GTMo_C_001_numpyro import kinetics_model as type1model
from HSA_GTMo_C_001_numpyro import TRUE_PARAMS as type1params
from HSA_GTMo_C_002_numpyro import kinetics_model as type2model
from HSA_GTMo_C_002_numpyro import TRUE_PARAMS as type2params
from HSA_GTMo_C_003_numpyro import kinetics_model as type3model
from HSA_GTMo_C_003_numpyro import TRUE_PARAMS as type3params
from HSA_GTMo_C_004_numpyro import kinetics_model as type4model
from HSA_GTMo_C_004_numpyro import TRUE_PARAMS as type4params
```

```python
import numpy
import jax
import jax.numpy as jnp
import matplotlib.pyplot as plt
import ICIW_Plots
from ipywidgets import IntSlider, interact
```

```python
ts = jnp.linspace(0, 12, 10)
```

```python
Ts = jnp.linspace(450, 600, 25, endpoint=True).reshape(-1, 1)
ps = jnp.full_like(Ts, 6.0)
c2hs = jnp.full_like(Ts, 1.0)
c2ms = jnp.full_like(Ts, 1.0)
c2ws = jnp.full_like(Ts, 1.0)
c2is = jnp.full_like(Ts, 100.0)
pi = calculate_partial_pressures(ps, c2hs, c2ms, c2ws, c2is)
sol1 = type1model(pi, Ts, *type1params.values())
sol2 = type2model(pi, Ts, *type2params.values())
sol3 = type3model(pi, Ts, *type3params.values())
sol4 = type4model(pi, Ts, *type4params.values())
```

```python
@interact(
    i=IntSlider(
        min=0,
        max=25,
        value=0,
        step=1,
    )
)
def interactive(i):
    plt.plot(ts, sol1[:, i, :, :].squeeze(), label="Type 1")
    plt.plot(ts, sol2[:, i, :, :].squeeze(), label="Type 2")
    plt.plot(ts, sol3[:, i, :, :].squeeze(), label="Type 3")
    plt.plot(ts, sol4[:, i, :, :].squeeze(), label="Type 4")
    plt.legend()
```

```python
plt.plot(Ts, sol1[-1, :, :, :].squeeze(), label="Type 1")
plt.plot(Ts, sol2[-1, :, :, :].squeeze(), label="Type 2")
plt.plot(Ts, sol3[-1, :, :, :].squeeze(), label="Type 3")
plt.plot(Ts, sol4[-1, :, :, :].squeeze(), label="Type 4")
plt.legend()
```
